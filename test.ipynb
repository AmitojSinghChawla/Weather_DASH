```python
import pandas as pd
import requests

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 52.52,
    "longitude": 13.41,
    "hourly": "temperature_2m,relative_humidity_2m,wind_speed_10m"
}
r = requests.get(url, params=params)
data = r.json()

# data["hourly"] is {"time": [...], "temperature_2m": [...]}
# two equal-length arrays -> perfect for a DataFrame
dataset = pd.DataFrame(data["hourly"])
dataset['time'] = pd.to_datetime(dataset['time'],unit="s",utc=True)
dataset.head()
```
